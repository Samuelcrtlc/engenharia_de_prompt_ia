#** Calculadora simples **

```python
def adicao(x, y):
    return x + y

def subtracao(x, y):
    return x - y


def multiplicacao(x, y):
    return x * y

def divisao(x, y):
    if y == 0:
        raise ValueError("Erro: Divisão por zero não é permitida.")
    return x / y

def calculadora():
    print("\n--- Calculadora Simples ---")
    print("Operações disponíveis: +, -, *, /")
    print("Para sair, digite 'sair'")

    while True:
        try:
            # Solicita o primeiro número ao usuário
            num1_str = input("Digite o primeiro número (ou 'sair'): ")
            # Verifica se o usuário deseja sair
            if num1_str.lower() == 'sair':
                break
            # Converte a entrada para um número de ponto flutuante
            num1 = float(num1_str)

            # Solicita o operador ao usuário
            operador = input("Digite o operador (+, -, *, /): ")
            # Verifica se o usuário deseja sair
            if operador.lower() == 'sair':
                break
            # Valida o operador inserido
            if operador not in ('+', '-', '*', '/'):
                print("Operador inválido. Por favor, tente novamente.")
                continue # Continua para a próxima iteração do loop

            # Solicita o segundo número ao usuário
            num2_str = input("Digite o segundo número (ou 'sair'): ")
            # Verifica se o usuário deseja sair
            if num2_str.lower() == 'sair':
                break
            # Converte a entrada para um número de ponto flutuante
            num2 = float(num2_str)

            # Realiza a operação com base no operador
            if operador == '+':
                resultado = adicao(num1, num2)
            elif operador == '-':
                resultado = subtracao(num1, num2)
            elif operador == '*':
                resultado = multiplicacao(num1, num2)
            elif operador == '/':
                resultado = divisao(num1, num2)

            # Exibe o resultado da operação
            print(f"Resultado: {num1} {operador} {num2} = {resultado}")

        # Captura erros de valor (ex: entrada não numérica, divisão por zero)
        except ValueError as e:
            print(f"Erro: {e}. Por favor, insira números válidos.")
        # Captura quaisquer outros erros inesperados
        except Exception as e:
            print(f"Ocorreu um erro inesperado: {e}")
        # Imprime um separador para melhor visualização
        print("---------------------------")

    # Mensagem de encerramento da calculadora
    print("Calculadora encerrada.")

# Chamada da função da calculadora para iniciar
calculadora()

# --- Sugestões para funções matemáticas adicionais e tratamento de erros ---
# Para adicionar mais funções matemáticas, você pode importar o módulo 'math' do Python.
# Exemplo:
# import math
#
# def raiz_quadrada(x):
#     if x < 0:
#         raise ValueError("Não é possível calcular a raiz quadrada de um número negativo.")
#     return math.sqrt(x)
#
# def potencia(base, expoente):
#     return math.pow(base, expoente)
#
# def seno(angulo_graus):
#     return math.sin(math.radians(angulo_graus)) # Converte graus para radianos

# Você também pode adicionar mais 'elif' na função 'calculadora' para incluir essas novas operações.
# Exemplo de como poderia ser na calculadora:
# elif operador == 'sqrt':
#     num = float(input("Digite o número para a raiz quadrada: "))
#     resultado = raiz_quadrada(num)
# elif operador == '^':
#     resultado = potencia(num1, num2)

# Para tratamento de erros, continue usando blocos `try-except` para capturar
# `ValueError` para entradas inválidas (não numéricas, por exemplo) ou lógicas (divisão por zero).
# Você pode ter `except` específicos para diferentes tipos de erro.
```
