### Simulação de Superaquecimento com Função Python

Este código define uma função `check_overheating` que verifica se uma temperatura está acima de um limite definido, indicando superaquecimento.

```python
def check_overheating(temperature, threshold=90):
    """
    Verifica se a temperatura está acima do limite de superaquecimento.

    Args:
        temperature (float or int): A temperatura atual.
        threshold (float or int): O limite de temperatura para superaquecimento.

    Returns:
        str: Uma mensagem indicando o status da temperatura.
    """
    if temperature > threshold:
        return f"ALERTA: Superaquecimento detectado! Temperatura: {temperature}°C (Limite: {threshold}°C)"
    else:
        return f"Temperatura normal: {temperature}°C (Limite: {threshold}°C)"

# Exemplo de uso da função
try:
    # Pede ao usuário para inserir a temperatura
    input_temp = float(input("Digite a temperatura atual para verificar (em °C): "))
    status_user_input = check_overheating(input_temp)
    print(status_user_input)
except ValueError:
    print("Entrada inválida. Por favor, digite um número para a temperatura.")

# Exemplo 1: Testando com uma temperatura abaixo do limite padrão (90°C)
current_temp_1 = 85
status_1 = check_overheating(current_temp_1)
print(status_1)
```
